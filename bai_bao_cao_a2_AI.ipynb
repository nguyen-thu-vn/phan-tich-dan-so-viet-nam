```python

import pandas as pd
import matplotlib.pyplot as plt

# Sắp xếp dữ liệu theo năm
df_urban = df_vn.sort_values("Nam").copy()

# Dân số thành thị và tổng dân số (đơn vị: nghìn người)
df_urban["Dan_so_thanh_thi"] = df_urban["Thanh_thi"]
df_urban["Dan_so_nong_thon"] = df_urban["Nong_thon"]

# Tỷ lệ đô thị hóa (%)
df_urban["Ty_le_do_thi_hoa"] = (
    df_urban["Dan_so_thanh_thi"] / df_urban["Dan_so"] * 100
)

# Tốc độ tăng dân số toàn quốc và thành thị (%/năm)
df_urban["Tang_dan_so_pct"] = (
    df_urban["Dan_so"].pct_change() * 100
)

df_urban["Tang_thanh_thi_pct"] = (
    df_urban["Dan_so_thanh_thi"].pct_change() * 100
)

# Mức thay đổi tỷ lệ đô thị hóa theo điểm phần trăm
df_urban["Tang_do_thi_hoa_diem_pct"] = (
    df_urban["Ty_le_do_thi_hoa"].diff()
)

# Bảng kết quả
cols = [
    "Nam",
    "Dan_so",
    "Dan_so_thanh_thi",
    "Ty_le_do_thi_hoa",
    "Tang_dan_so_pct",
    "Tang_thanh_thi_pct",
    "Tang_do_thi_hoa_diem_pct"
]

print("BẢNG PHÂN TÍCH ĐÔ THỊ HÓA VÀ TĂNG TRƯỞNG DÂN SỐ")
display(df_urban[cols].round(2))

# So sánh đầu kỳ và cuối kỳ
first = df_urban.iloc[0]
last = df_urban.iloc[-1]

growth_population = (
    (last["Dan_so"] / first["Dan_so"] - 1) * 100
)

growth_urban_population = (
    (last["Dan_so_thanh_thi"] / first["Dan_so_thanh_thi"] - 1) * 100
)

urbanization_start = first["Ty_le_do_thi_hoa"]
urbanization_end = last["Ty_le_do_thi_hoa"]

print("\nSO SÁNH GIAI ĐOẠN 1995–2024")
print(f"Tăng dân số toàn quốc: {growth_population:.2f}%")
print(f"Tăng dân số thành thị: {growth_urban_population:.2f}%")
print(f"Tỷ lệ đô thị hóa năm 1995: {urbanization_start:.2f}%")
print(f"Tỷ lệ đô thị hóa năm 2024: {urbanization_end:.2f}%")
print(
    "Mức tăng tỷ lệ đô thị hóa: "
    f"{urbanization_end - urbanization_start:.2f} điểm phần trăm"
)

# Biểu đồ 1: Tỷ lệ đô thị hóa qua thời gian
plt.figure(figsize=(12, 6))
plt.plot(
    df_urban["Nam"],
    df_urban["Ty_le_do_thi_hoa"],
    marker="o"
)
plt.title("Tỷ lệ đô thị hóa Việt Nam (1995–2024)")
plt.xlabel("Năm")
plt.ylabel("Dân số thành thị / Tổng dân số (%)")
plt.xticks(range(1995, 2025, 3), rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Biểu đồ 2: Tốc độ tăng dân số toàn quốc và dân số thành thị
plt.figure(figsize=(12, 6))
plt.plot(
    df_urban["Nam"],
    df_urban["Tang_dan_so_pct"],
    marker="o",
    label="Tăng dân số toàn quốc"
)
plt.plot(
    df_urban["Nam"],
    df_urban["Tang_thanh_thi_pct"],
    marker="o",
    label="Tăng dân số thành thị"
)
plt.title("So sánh tốc độ tăng dân số toàn quốc và thành thị")
plt.xlabel("Năm")
plt.ylabel("Tăng trưởng so với năm trước (%)")
plt.xticks(range(1996, 2025, 3), rotation=45)
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Tìm hàng tiêu đề thật trong DataFrame df
header_mask = df.astype(str).apply(
    lambda col: col.str.strip().eq("Tỉnh, thành phố / Vùng")
).any(axis=1)

if not header_mask.any():
    raise ValueError(
        "Không tìm thấy hàng tiêu đề. "
        "Hãy kiểm tra df.head(10) và df.columns."
    )

header_row = header_mask[header_mask].index[0]

# 2. Tạo DataFrame sạch
cleaned_df = df.iloc[header_row + 1:].copy()
cleaned_df.columns = (
    df.iloc[header_row].astype(str).str.strip().values
)
cleaned_df = cleaned_df.reset_index(drop=True)
cleaned_df.columns = cleaned_df.columns.str.strip()

print("Các cột tìm được:")
print(cleaned_df.columns.tolist())

# 3. Xác định tên cột mật độ dân số
density_col = next(
    (
        col for col in cleaned_df.columns
        if "Mật độ dân số" in col
    ),
    None
)

year_col = next(
    (col for col in cleaned_df.columns if col.strip() == "Năm"),
    None
)

region_col = next(
    (
        col for col in cleaned_df.columns
        if "Tỉnh, thành phố" in col or "Vùng" in col
    ),
    None
)

if not all([year_col, region_col, density_col]):
    raise ValueError(
        "Không tìm thấy đủ cột Năm, Vùng và Mật độ dân số. "
        f"Các cột hiện có: {cleaned_df.columns.tolist()}"
    )

# 4. Chuyển dữ liệu về kiểu phù hợp
cleaned_df[year_col] = pd.to_numeric(
    cleaned_df[year_col], errors="coerce"
)

cleaned_df[density_col] = pd.to_numeric(
    cleaned_df[density_col]
        .astype(str)
        .str.replace(",", "", regex=False)
        .str.strip(),
    errors="coerce"
)

cleaned_df[region_col] = (
    cleaned_df[region_col].astype(str).str.strip()
)

cleaned_df = cleaned_df.dropna(
    subset=[year_col, density_col]
)

cleaned_df = cleaned_df[
    cleaned_df[year_col].between(1995, 2024)
].copy()

# 5. Chọn 6 vùng kinh tế - xã hội
regions = [
    "Đồng bằng sông Hồng",
    "Trung du và miền núi phía Bắc",
    "Bắc Trung Bộ và Duyên hải miền Trung",
    "Tây Nguyên",
    "Đông Nam Bộ",
    "Đồng bằng sông Cửu Long"
]

region_df = cleaned_df[
    cleaned_df[region_col].isin(regions)
].copy()

if region_df.empty:
    print("Không tìm thấy tên vùng khớp với danh sách.")
    print("Các tên vùng/địa phương trong dữ liệu:")
    print(sorted(cleaned_df[region_col].unique()))
else:
    # 6. Vẽ biểu đồ đường
    sns.set_theme(style="whitegrid")
    plt.figure(figsize=(15, 8))

    sns.lineplot(
        data=region_df,
        x=year_col,
        y=density_col,
        hue=region_col,
        marker="o",
        errorbar=None
    )

    plt.title(
        "Mật độ dân số theo vùng tại Việt Nam (1995–2024)",
        fontsize=15
    )
    plt.xlabel("Năm")
    plt.ylabel("Mật độ dân số (người/km²)")
    plt.legend(
        title="Vùng",
        bbox_to_anchor=(1.02, 1),
        loc="upper left"
    )
    plt.grid(True, linestyle="--", alpha=0.4)
    plt.tight_layout()
    plt.show()
```
